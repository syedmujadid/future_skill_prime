# Exercise 1 · Exploratory Data Analysis and Automated Reporting
Generate sample data with NumPy, explore it with a scatter plot, calculate row means with Pandas, and export the result to Excel.

Run the cells from top to bottom. This notebook uses numpy, pandas, matplotlib, and openpyxl.

# 1. Generate random data
Create 100 rows of random integers from 1 to 100. The seed keeps the sample the same each time you run the notebook. df.head() previews the first five rows.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=42)
columns = ["f1", "f2", "f3", "f4"]
values = rng.integers(1, 101, size=(100, len(columns)))
df = pd.DataFrame(values, columns=columns)
df.head()
```

The fixed seed makes the sample reproducible. Each point below represents one row,
with `f2` on the horizontal axis and `f3` on the vertical axis.

# 2. Plot f2 against f3
Each dot represents one row. Its horizontal position is the value of f2 and its vertical position is the value of f3.

```python
fig, ax = plt.subplots(figsize=(7, 4.5))
ax.scatter(df["f2"], df["f3"], color="#156879", alpha=0.75)
ax.set(xlabel="f2", ylabel="f3", title="Scatter plot of f2 and f3")
ax.grid(alpha=0.2)
ax.set_axisbelow(True)
fig.tight_layout()
plt.show()
```
