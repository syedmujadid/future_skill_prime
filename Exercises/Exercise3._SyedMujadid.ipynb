### Keyframes and timing

Exact prompt
Generate a 3-second video from the four supplied keyframes. Use a locked, straight-on camera and keep the full composition visible throughout. From 0 to 1 second, smoothly cross-dissolve from the yellow-circle image to the teal-rectangle image. From 1 to 2 seconds, smoothly cross-dissolve to the supplied blended image, with both shapes visible and their overlap matching the reference. From 2 to 3 seconds, hold the blended image completely still. Keep the original colors, shape edges, positions, and proportions. Use opacity changes only. Avoid pans, tilts, zooms, cropping, shape morphing, flicker, text, and extra objects.

Set the transition length to 1 second. Four keyframes give three transitions and a total duration of 3 seconds.
