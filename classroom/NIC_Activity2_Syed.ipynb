```python
# Creating a dataset of popular Indian singers and their achievements
data = {
    'Singer': ['Arijit Singh', 'Shreya Ghoshal', 'Sonu Nigam', 'Lata Mangeshkar', 'Kishore Kumar'],
    'Years_Active': [15, 20, 30, 70, 40],
    'Number_of_Songs': [500, 700, 600, 1000, 800],
    'Awards_Won': [50, 60, 40, 100, 80]
}

# Convert the dictionary into a Pandas DataFrame
singers_df = pd.DataFrame(data)

# Display the dataset
singers_df
```

```python
#Display basic information about dataset
singers_df.info()
```

```python
#show the first few rows of the dataset

#for last records
singers_df.tail(2)

#for first records
singers_df.head(2)
```

```python
#sort singers by numbers of songs

sorted_singers = singers_df.sort_values(by='Awards_Won', ascending=False)
sorted_singers
```

```python
import matplotlib.pyplot as plt

# Bar chart of the number of songs by each singer
plt.figure(figsize=(10, 5))
plt.bar(singers_df['Singer'], singers_df['Number_of_Songs'], color='green')
plt.title('Number of Songs by Indian Singers')
plt.xlabel('Singer')
plt.ylabel('Number of Songs')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

```python
Looking only at f_01 through f_04, notice how when f_01 is near 0.50, f_03 goes up near 100 and f_02 drops low. Without knowing what we are predicting, what do you think these sensors are measuring?
f_01 – Normalized Time of Day (0.00 = Midnight, 0.50 = Noon, 0.95 = ~11 PM). Values near 0.50 correspond to peak daytime conditions.
f_02 – Relative Humidity (proportion), inversely related to f_03 and temperature (drier air during hotter midday hours).
f_03 – Solar / Sunlight Index, strongly correlated with midday (f_01 near 0.50) and higher temperatures.
f_04 – Location / Zone Type (A = Urban/Concrete, B = Park/Green Zone). Categorical feature. “A” rows run ~2–3°C warmer than “B” rows under similar conditions (urban heat island effect).
Target_Temp_C – Ambient Temperature (°C) (Target Label)
```
