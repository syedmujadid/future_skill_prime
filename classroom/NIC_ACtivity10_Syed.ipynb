```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread(r"nic.PNG")
```

```python
print(type(img))
```

```python
print(img.shape)
```

```python
if img is None:
    print("Error: Could not load image. Check the file path and name!")
else:
    # First argument must be the window name string, second is the image array
    print("showing image")
    cv2.imshow('Bootcamp Image', img) 
    cv2.waitKey(0)
```

```python
cv2.destroyAllWindows()
```

```python
img_resize = cv2.resize(img, (100, 100), interpolation=cv2.INTER_AREA)
```

```python
cv2.imshow("My Image",img_resize)
```

```python
cv2.destroyAllWindows()
```

```python
img_flip = cv2.flip(img,1)
```

```python
cv2.imshow("Bootcamp",img_flip)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
